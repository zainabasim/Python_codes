# 📘 Python Lists in Colab with Outputs & Uses

## 1. Creating Lists

```python
# Creating lists
empty = []
numbers = [1, 2, 3, 4, 5]
mixed = ["Ali", 42, True, [1, 2, 3]]

print("Empty:", empty)
print("Numbers:", numbers)
print("Mixed:", mixed)
```

## 2. Accessing Elements

```python
fruits = ["apple", "banana", "cherry"]

print("First:", fruits[0])
print("Last:", fruits[-1])
```

## 3. Adding Elements

```python
fruits = ["apple", "banana", "cherry"]

fruits.append("date")          # add at the end
fruits.insert(1, "blueberry")  # add at index 1

print("After adding:", fruits)
```

## 4. Removing Elements

```python
fruits = ["apple", "banana", "cherry", "banana"]

removed_last = fruits.pop()     # removes last
fruits.remove("banana")         # removes first "banana"

print("After removal:", fruits)
print("Last removed:", removed_last)
```

## 5. Finding and Counting

```python
fruits = ["apple", "banana", "cherry", "banana"]

print("Index of cherry:", fruits.index("cherry"))
print("Banana count:", fruits.count("banana"))
```

## 6. Extending a List

```python
fruits = ["apple", "banana", "cherry"]
more_fruits = ["elderberry", "fig"]

fruits.extend(more_fruits)
print("Extended list:", fruits)
```

## 7. Reversing a List

```python
fruits = ["apple", "banana", "cherry"]

fruits.reverse()
print("Reversed list:", fruits)
```

## 8. Sorting a List

```python
fruits = ["banana", "apple", "cherry", "date"]

fruits.sort()                 # ascending
print("Ascending:", fruits)

fruits.sort(reverse=True)     # descending
print("Descending:", fruits)
```
